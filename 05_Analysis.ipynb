```python
import numpy as np
import pickle
from eval_tools import getGroundTruthTimestamps
import plotly.graph_objs as go
```

# Plot Alignments

```python
s = 6
modes = ['random', 'continuous', 'constant']
mode_id = 0
oltw_hyp_file = f"experiments/{modes[mode_id]}//OLTW/s{s}/hyp.npy"
oltw_tsm_file = f"experiments/{modes[mode_id]}/OLTW/s{s}/tsm.npy"
dtw_hyp_file = f"experiments/{modes[mode_id]}/NaivePairwiseDTW/s{s}/hyp.npy"
dtw_tsm_file = f"experiments/{modes[mode_id]}/NaivePairwiseDTW/s{s}/tsm.npy"
dtw_tsm_lag_400_file = f"experiments/{modes[mode_id]}/NaivePairwiseDTW/s{s}/tsm_lag400.npy"

# load the hyp and tsm files
match_hyp = np.load(oltw_hyp_file)
match_tsm = np.load(oltw_tsm_file)
dtw_hyp = np.load(dtw_hyp_file)
dtw_tsm = np.load(dtw_tsm_file)
dtw_tsm_lag_400 = np.load(dtw_tsm_lag_400_file)

# load the ground truth timestamps
query_annot_file = f'scenarios/{modes[mode_id]}/s{s}/p.beats'
ref_annot_file = f'scenarios/{modes[mode_id]}/s{s}/o.beats'
scenarioInfo = f'scenarios/{modes[mode_id]}/s{s}/scenario.info'
gt = getGroundTruthTimestamps(query_annot_file, ref_annot_file, scenarioInfo)
```

```python
gt = gt[0].T
```

```python
# Create a figure
fig = go.Figure()
fig.update_layout(
    title='Alignment and TSMvPaths',
    xaxis_title='Query Time (s)',
    yaxis_title='Reference Time (s)',
    width=900,
    height=700
)
fig.add_trace(go.Scatter(
    x=match_hyp[0], y=match_hyp[1], mode='lines', name='OLTW Hyp'
))
fig.add_trace(go.Scatter(
    x=match_tsm[0], y=match_tsm[1], mode='lines', name='OLTW TSM'
))
fig.add_trace(go.Scatter(
    x=dtw_hyp[0], y=dtw_hyp[1], mode='lines', name='DTW Hyp'
))
fig.add_trace(go.Scatter(
    x=dtw_tsm[0], y=dtw_tsm[1], mode='lines', name='DTW TSM'
))
fig.add_trace(go.Scatter(
    x=gt[0], y=gt[1], mode='markers', name='Ground Truth'
))
fig.show()
```

# Plot Errors

```python
oltw_eval_file = f"eval/{modes[mode_id]}/OLTW/lag0/errs.pkl"
oltw_tsm_eval_file = f"eval_tsm/{modes[mode_id]}/OLTW/lag400/errs.pkl"
# dtw_eval_file = f"eval/{modes[mode_id]}/NaivePairwiseDTW/lag0/errs.pkl"
# dtw_tsm_eval_file = f"eval_tsm/{modes[mode_id]}/NaivePairwiseDTW/lag0/errs.pkl"
# dtw_tsm_lag_400_eval_file = f"eval_tsm/{modes[mode_id]}/NaivePairwiseDTW/lag400/errs.pkl"

oltw_err = pickle.load(open(oltw_eval_file, 'rb'))[f's{s}']
oltw_tsm_err = pickle.load(open(oltw_tsm_eval_file, 'rb'))[f's{s}']
# dtw_err = pickle.load(open(dtw_eval_file, 'rb'))[f's{s}']
# dtw_tsm_err = pickle.load(open(dtw_tsm_eval_file, 'rb'))[f's{s}']
# dtw_tsm_lag_400_err = pickle.load(open(dtw_tsm_lag_400_eval_file, 'rb'))[f's{s}']
```

```python
# plot the errors
fig = go.Figure()
fig.update_layout(
    title='Alignment and TSM Errors',
    xaxis_title='Reference Time (s)',
    yaxis_title='Error (s)',
    yaxis=dict(range=[-2, 2])
)
fig.add_trace(go.Scatter(x=np.arange(len(match_err)), y=match_err, mode='lines', name='OLTW'))
fig.add_trace(go.Scatter(x=np.arange(len(match_tsm_err)), y=match_tsm_err, mode='lines', name='OLTW TSM'))
fig.add_trace(go.Scatter(x=np.arange(len(dtw_err)), y=dtw_err, mode='lines', name='DTW'))
fig.add_trace(go.Scatter(x=np.arange(len(dtw_tsm_err)), y=dtw_tsm_err, mode='lines', name='DTW TSM'))
fig.show()
```
